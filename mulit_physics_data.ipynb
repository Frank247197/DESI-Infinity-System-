```python
import numpy as np
import pandas as pd

num_samples = 10000
np.random.seed(42)

# 1. Conduction Data: Q = k * A * dT / L
k = np.random.uniform(0.1, 400.0, num_samples)
A_c = np.random.uniform(0.01, 1.0, num_samples)
dT_c = np.random.uniform(1.0, 500.0, num_samples)
L = np.random.uniform(0.05, 0.5, num_samples)
Q_conduction = (k * A_c * dT_c) / L

print(f"Generated {num_samples} samples for Conduction.")
print(f"Sample Conduction Q (W): {Q_conduction[0]:.2f}")
```

```python
# 2. Convection Data: Q = h * A * dT
h = np.random.uniform(5.0, 5000.0, num_samples)
A_v = np.random.uniform(0.01, 5.0, num_samples)
dT_v = np.random.uniform(1.0, 300.0, num_samples)
Q_convection = h * A_v * dT_v

print(f"Generated {num_samples} samples for Convection.")
print(f"Sample Convection Q (W): {Q_convection[0]:.2f}")

# Combine into a single report and save
df = pd.DataFrame({
    'k': k, 'A_c': A_c, 'dT_c': dT_c, 'L': L, 'Q_conduction': Q_conduction,
    'h': h, 'A_v': A_v, 'dT_v': dT_v, 'Q_convection': Q_convection
})
df.to_csv('combined_heat_data.csv', index=False)
print("Data saved successfully!")
```

# Multi-Physics Heat Transfer Dataset

This notebook generates 10,000 randomized samples for two fundamental modes of heat transfer:

### 1. Conduction (Fourier's Law)
**Formula:** Q = (k * A * dT) / L
* k = thermal conductivity (W/mK)
* A = cross-sectional area (m^2)
* dT = temperature difference (K)
* L = thickness/length (m)

### 2. Convection (Newton's Law of Cooling)
**Formula:** Q = h * A * dT
* h = convective heat transfer coefficient (W/m^2K)
* A = surface area (m^2)
* dT = temperature difference (K)

The goal is to train a Neural Network to learn both physical laws simultaneously from this combined dataset.
